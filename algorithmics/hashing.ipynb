# Hashing

Our goal is to manage dictionaries efficiently. Let $S\subset U$ bet the set of keys and $U$ the universe to which this set belongs. We use a so called *Hash Function* to describe keys. **Hash function** orders each element of the universe $U$ to number in $\{0,..,m-1\}$ $$h_m=h:U\longrightarrow\{0,...,m-1\}$$The set of keys $S$ can be big, which means $$B_i = \{x \in S: h(x) =i\} \ \ \ \ i = 0,1,...,m-1$$ are often large sets, called **buckets**. The cases of a bucket containing more than one element are called **collisions**.

# Collisions

We define **load factor** of a hash table as $\alpha := \frac{n}{m}$

We want to study probability of collisons. For this, we make the following assumptions:
1. $n\leq m << |U|$. (The are more buckets than keys, the universe is large)
2. $\alpha_0<\alpha\leq1$, where $a_0\in(0,1)$. (There's no space to waste, $n\leq m$)
3. $n << |U_i|$. All keys can be picked from a bucket.

Keys from the universe $U$ are picked as random elements $X_0,...,X_{n-1}$.


### Uniformity Assumption (UF)
1.  $i\neq j \implies X_i \neq x_j$
2.  $h(X_0),...,h(X_{n-1})$ is a sequence of *independent* random variables from ${0,...,m-1}$ with $P(h(x_i)=j)=\frac{1}{m},$ for all $ j\in \{0,...,m-1\}$
  

In short:

$$\text{(UF):}\implies \text{ no two keys are the same}$$

$$\implies \text{values are random}$$

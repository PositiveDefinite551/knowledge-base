# Heap



#### Definition:

A heap is a binary tree with the two properties:
- v is a parent of w $\implies \text{key}(v) \leq \text{key}(w)$
- It's complete up to the lowest level, which is also complete from the left to the right up to a certain place.

Labeling each node from up to down, from left to right allows us to implement Heap as array $A[o,n)$ of integers.



#### Properties
Let i be a node of a heap A, with $\text{heapsize}=n$ and $i\in \{0,1,...,n-1\}$. We observe the three scenarious:
1. $n \leq 2i+1 \implies$ i has no children
2. $n-1 = 2i+1 \implies$ i has exactly one child
3. $2i+2\leq n-1 \implies$ i has exactly two children.

We describe a heap using indizes:
- left child of $i$ is $2i+1$
- right child of $i$ is $2i+2$
- parent of $i$ is $\text{floor}(\frac{i-1}{2})$

$\text{depth}A = \text{floor}(\log_2n)$


# Turning array into a heap.




#### Heapify
We implement an algorithm `heapify`, which repairs the heap, assuming only the node $k$ is misplaced. It has following properties:

$\text{keyComp}(\text{heapify}(k, l)) \leq 2 \cdot \text{floor}(\log _2 \frac{l}{k+1})$

$\text{time}(\text{heapify}(k,l))= O(\log \frac{l}{k+1})$


```python
def heapify(A, k, l):
    """
    Takes node at the index k and if it's violating the heap property puts it into the right place.
    Args:
        A (List[int]]) : The array to work on.
        k (int) : the node to heapify.
        l (int) : the end of the range to work on. 
    
    """
    if 2*k+1 >= l:
        return
    elif 2*k+1 == l-1:
        maxchild = 2*k + 1
    else: 
        if A[2*k+1] > A[2*k+2]:
            maxchild = 2*k + 1
        else: 
            maxchild = 2*k + 2
        
    A[k], A[maxchild] = A[maxchild], A[k]
    heapify(A, maxchild, l)
    
```

Now we can build a heap by calling heapify on each node from the bottom up. We skip the nodes that are placed on the last level:

```python
def buildheap(A):
    n = len(A)
    for j in range(round(n/2)-1, -1, -1):
        heapify(A, j, n)
```

#### BuildHeap
The algorithm buildheap has following Properties:
$\text{keyComp}(\text{buildheap}(A) \leq 2n$

$\text{time}(\text{buildheap}(A))= O(n)$

where $n$ is the length of $A$.

```python
A = [1,2,3,4,5]
buildheap(A)
A
```

# Heapsort
Using heaps for sorting an array.
The algorithm has following Properties:

$\text{keyComp}(\text{heapsort}(A) \leq 2n(\text{floor}(\log _2 n)+1)$

$\text{time}(\text{heapsort}(A))= O(n \log n)$


```python
def heapsort(A):
    """
        Sorts an array of integers in-place using HeapSort.
        Args:
            A List[int] : The array of integers to sort
        
    """
    n = len(A)
    if n <= 1:
        return
        
    buildheap(A)
    
    heapsize = n
    while heapsize >= 2:
        A[0], A[heapsize - 1] = A[heapsize - 1], A[0]
        heapsize = heapsize - 1
        heapify(0, heapsize)
```

#### Interpretation
Essentialy, the algorithm prepares the array into the suitable structure in just $n$ steps. Then it looks for the biggest element $n$ times. Each such search requires only logarithmic amount of steps, thanks to the heap structure.

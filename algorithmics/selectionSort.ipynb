# SelectionSort

KeyComp $Θ(n^2)$

Time $O (n^2)$

**Very important**: To sort an array with $n$ element, the `SelectionSort` has to make AT MOST $n$ key exchanges. **Lots of key comparisons, few Key exchanges**.


#### Principe
- Split the array into two parts S and U.
- S will be always sorted and U unsorted.
- in the beginning S is empty and U is all the array.
- Until U isn't empty, swap the smallest element of U for its first and move theborder of S to include this element and that of S to exclude.


#### Runtime
- The minimum element of the array has to be found $n-1$ times
- The element has to be swapped $n-1$ times as well.

# Search Trees


Let a search tree be given. We define tree traversals by the sequence of nodes visited.

- **Preorder** : Root, Left, Right

- **Postorder** : Left, Right, Root

- Only for binary trees: **Inorder** : Left, Root, Right 

# Binary Search Tree

We want to implement a binary search tree. For this we make a class `Node`, which will save the key, left and right child.

```python
class Node:
    def __init__(self, key, data=None):
        self.key = key
        self.lchild = None
        self.rchild = None
        self.parent = None
        self.data = data

    def insert_child(self, k):
        if k > self.key:
            if self.rchild==None:
                self.rchild = Node(k)
                self.rchild.parent = self
                return
            else:
                self.rchild.insert_child(k)
            
        else: 
            if k < self.key and self.lchild==None:
                self.lchild = Node(k)
                self.lchild.parent = self
                return
            else:
                self.lchild.insert_child(k)

class BinaryTree:
    def __init__(self, root):
        self.root = root

    def insert(self, key):
        self.root.insert_child(key)

def preorder(root):
    if root == None:
        return []
    return [root.key] + preorder(root.lchild) + preorder(root.rchild)

def postorder(root):
    if root == None:
        return []
    return postorder(root.left) + postorder(root.right) + root.key

def inorder(root):
    if root == None:
        return []
    return inorder(root.left) + [root.key] + inorder(root.right)

```

### Greatest and Least
- `greatest(u)`. The greatest element of the symmetrical order of nodes of a tree can be acquired by following the right child of each node starting from the root, until there's none left.
-  `least(u)`. *Replace 'greatest' with 'smallest' and 'right' with 'left' (except for the last word :D)*

We implement the following algorithms. Both of them have a runtime complexity of $O(\text{depth}T_u)$

```python
def greatest(u : Node) -> int:
    if u.rkind==None:
        return u.key
    return greatest(u.rkind)

def least(u):
    if u.lkind==None:
        return u.key
    return least(u.lkind)
```

 # Previous and Next
Let T be a nonempty BST, $v\in T$. There are exactly two cases possible:

**Case 1**: v has no predecessor ($\implies$ v is the smallest node). It is true if and only if:
- v has no left child
- v is a root or none of the nodes in the path of v are right children of their parents.

**Case 2**: v has a predecessor p. Let $T_u:=T_{v,p}$. Then there are two subcases:
1. $v=u$ and p is the maximal element of the left subtree of u. It's true if and only if v has a left child.
2. $p=u$ and v is the minimal node of the right subtree of u.

We implement the following algorithms to find a previous element of a node. Each of them has a complexity $$O(\text{depth}T)$$ where T is the tree to which the nodes belong.

```python
def previous(u : Node) -> Node:
    """
        Assumption: u isn't the smallest node of the tree.
    """
    if u.lchild!=None:
        return greatest(u.lchild)
    w = u
    p = u
    while w != p.rchild:
        w = p
        p = p.parent
    return p
        
```

```python
def next(u : Node) -> Node:
    """
        Assumption: u isn't the greatest node of the tree.
    """
    if u.rchild!=None:
        return least(u.rchild)

    w = u
    p = u
    while p.lchild != w:
        w = p
        p = p.parent
    return p
```

# Search
runtime: $O(\text{depth}T)$

```python
def searchPath(k : int, v : Node) -> Node:
    """
        Calculates the last node in the search path of a given integer..
        Args:
            k (int) : The key to search for.
            v (Node) : the root of the tree to search
    """
    case1 = v.key == k #k belongs to the root
    case2 = v.key < k and v.rkind == None #k belongs to the right of the root but it's empty
    case3 = v.key > k and v.lkind == None #k belongs to the right of the root but it's empty
    if case1 or case2 or case3:
        return v

    # if k isn't the root and the root has a child, continue search:
    if k>v.key:
        return searchPath(k, v.rchild)
    if k<v.key:
        return searchPath(k, v.lchild)
    
    
```

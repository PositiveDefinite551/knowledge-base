# Basic notions of complexity

### Random Access Machine (RAM)
Consist of memory cells, which can be read and written to via instructions.

Each memory cell can entail $\omega$ bits, which means the model allows us to work on input size $n$ where $\omega > \log n$. This ensures each memory cell is large enough to hold a pointer to any position in the input.

#### Basic operations
RAM can execute the following basic operations in exactly one step:
- Arithmetic operations and comparing numbers
- Data movement (indirect adressing, saving  a number in a cell, load the content of a cell in a variable, copy content from one cell to another)
- Control structures (conditional statemets, Branching/Jumps, Subprogram Call/Return).
All of the operatoins are executed sequentially

#### Runtime
Runtime of an algorithm is a numer of Basic operations that RAM executes while running it. Let $A$ be an algorithm and $I$ the input. We define: 
- $\text{time}_A(I) = $ The number of basic operations that $A$ performs on $I$
- $\text{time}_A(n) = \sup \{\text{time}_A(I) : |I|=n \}$ (in the worst case)
- $\overline{\text{time}_A(n)} := \frac{1}{\pi _n} \sum_{I: |I|=n} \text{time}_A(I)$ (on average)

An algorithm is often called efficient when $\text{time}_A(n)= O(n^k)$ for $k\in\mathbb{N}$

#### Space
- $\text{space}_A(I) = $ The number of memory cells that are used when performing $A$ on $I$
- $\text{space}_A(I) = \sup \{\text{space}_A(I) : |I|=n \}$ (in the worst case)

### Decision Problem
Decision problem is a function $$f: \Sigma ^* \to \{0,1\},$$ where $\Sigma ^* $ is a set of words over the alphabet $\Sigma$ is. It is decided whether an input word $w\in \Sigma^*$ belongs to the formal language $$L:-f^{-1}(1)\subset \Sigma ^* $$

A **decision algorithm** is an algorithm that solves a decision problem.

### P vs. NP

**P**: The complexity class **P** consists of all the decision problems, that can be solved using a polynomial-time algorithm.

**NP**: The complexity class **NP** consists of all the decision problems, that can be solved using a **NP**-algorithm.

**NP**-decision algorithm is built as follows:
- Rate-Phase: Rate a certificate
- Verification phase: Is the given certificate an answer for the problem?
Is there a posigive certificate for an input? if yes, return 1, else 0.

Both of these phases consist of polynomial number of basic operations.

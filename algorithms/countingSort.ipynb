# Counting Sort and Radix Sort

## Counting Sort
Sort n whole numbers from the interval $[0,k)$

**time** = $O(n+k)$

### The idea

- For each number from the interval $[0,k)$ count how many times it shows up in the array to sort and store this info in the additional array `C[0,k]`.
- Increment each element of C by its previous element. Now C records the position at which according keys end.
- Iterate from back to the beginning through A, adding each element into a new (output) array B at the index `C[key of the element]` and lower `C[key of the element]` by one.

The last step also ensures that the Algorithm is stable.

### Python Implementation

```python
import numpy as np

class Data:
    def __init__(self, key, value=None):
        self.key = key
        self.value = value

    def __repr__(self):
        return str((self.key, self.value))

def countingSort(A : np.ndarray, k : int) -> np.ndarray:
    n = A.shape[0]
    B = np.zeros(n, dtype=object)
    C = np.zeros(k, dtype=int)

    for j in range(n):
        C[A[j].key] += 1 

    for j in range(1, k):
        C[j] = C[j-1] + 1 

    for j in range(n-1,-1,-1):
        B[C[A[j].key] - 1] = A[j]
        C[A[j].key] -=1

    return B
    
print(countingSort(np.array([Data(4), Data(7), Data(1), Data(4, "but this comes second"), Data(2), Data(4, "but this comes last") ]), 8))
```

## Radix Sort

$O(d(n+k))$

Let an array of $n$ elements be given, each of those consist of $d$ entries, which belong to the universe of length $k$.

for r=d to r=0 apply counting Sort on the array based on the r-th element.
Since CountingSort is stable, this will result in sorted array.

### QR -Decomposition

Let a $A \in \mathbb{K}^{m\times n}$ be given. Then there exists a
- **unitary** Matrix $Q \in \mathbb{K}^{m\times m} $
- generalised upper diagonal matrix $R \in \mathbb{K}^{m\times n} $,
  
such that $A = QR$. Because of $Q^{-1}=Q^{*}$, this would turn an equation system $Ax=b$ into $$Rx=Q^*b$$
where $Q^*b$ is the conjugate compose of $Q$. 

One way to calculate this decomposion is to orthogonalise the matrix $A$ using the Gram-Schmidt's algorithm, and record the matrix $R$ along the way. 

#### The algorithm

Let's say we alrady orthogonalised $k-1$ columns of A. That is we have $k-1$ vectors $q_0,...,q_{k-2}$, such that 
- $q_i*q_j=\delta _{i,j}$ and
- $\text{span}(q_0,...,q_{k-2}) = \text{span}(a_0,...,a_{k-1})$.

Now, to calulate the $k$-th orthogonal vector $q_k$ from the $k$-th column of A $a_k$, we subtract from $a_k$ the following linear compination that we get from the already calculated vectors and mark it $\hat{q}$: $$\hat{q} = a_k - \sum_{i=1}^{k-1}(a_k^*q_i)q_i$$ 
We also rewrite this equation into: $$=||\hat{q}_k||_2q_k + \sum_{i=1}^{k-1}(a_k^*q_i)q_i$$ and save the first summand as the $k$-th diagonal element of $R$, that is $r_{k,k}$ and each summand inside the iterated sum (depending on i) as $r_{i,k}$.

We then normalize the result for $\hat{q}$ and receive the new orthogonal vector: $$q_k:=\frac{\hat{q}_k}{||{\hat{q}_k}||_2}$$



This process gives us the first $n$, orthogonal vectors, which are enough to solve the system. The rest of the vectors can be obtained by completing $q_j$ to the orthogonal Basis.

The python implementation looks as follows:

```python
import numpy as np


def qr_decomposition(A : np.ndarray ) -> tuple[np.ndarray] :
    
    """
    Gram-Schmidt Algorithm for calculating the QR-decomposition.
    ---
    Args: 
        A(np.ndarray) : The mxn matrix to decompose
    ---
    Returns:
        tuple[np.ndarray] : Orthogonal matrix Q, upper-diagonal nxn matrix R, such that QR=A.
    """

    m,n = A.shape[0], A.shape[1]
    R = np.zeros(shape=(m,n))
    Q =  np.zeros(shape=(m,n)) # Here we will calulate the "reduced" Q

    for k in range(n):
        for i in range(k): # when k=1, this loop isn't executed
            R[i][k]= (np.transpose(np.conjugate(A[:,k]))) @ Q[:,i]
            
        sigma = 0
        for i in range(k):
            sigma += (np.transpose(np.conjugate(A[:,k])) @ Q[:,i] ) * Q[:,i]
            
        q_hat = A[:,k] - sigma
        R[k][k] = np.linalg.norm(q_hat)
        Q[:,k] = q_hat/R[k][k]

    return Q, R

```
